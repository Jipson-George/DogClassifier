```python
!pip install flask pyngrok tensorflow

import tensorflow as tf
from flask import Flask, request, jsonify
import numpy as np
from PIL import Image
import io
import base64
from pyngrok import ngrok
import threading
ngrok.set_auth_token("2wLY9k98Mg3bGaINGVg7Q6LCiyZ_4gbxFs8fcWzmKqJ8cqANf")
# Load your model
model = tf.keras.models.load_model("/improved_dog_classifier.h5")

# Create Flask app
app = Flask(__name__)

@app.route('/predict', methods=['POST'])
def predict():
    try:
        # Get image data
        data = request.get_json()
        image_data = base64.b64decode(data['image'])

        # Process image
        image = Image.open(io.BytesIO(image_data))
        image = image.resize((224, 224))  # ✅ match training size
        image_array = np.array(image) / 255.0
        image_array = np.expand_dims(image_array, axis=0)

        # Make prediction
        predictions = model.predict(image_array)
        score = float(predictions[0][0])


        if score >= 0.5:
            predicted_class = 1  # Not a Dog
            confidence = score
        else:
            predicted_class = 0  # Dog
            confidence = 1 - score

        return jsonify({
            'predicted_class': predicted_class,
            'confidence': confidence,
            'raw_score': score
        })


    except Exception as e:
        print(f"Error: {e}")
        return jsonify({'error': str(e)}), 500

@app.route('/health', methods=['GET'])
def health():
    return jsonify({'status': 'healthy'})

# Start ngrok tunnel
public_url = ngrok.connect(5000)
print(f"🌐 Public URL: {public_url}")

# Start Flask app in background
def run_app():
    app.run(host='0.0.0.0', port=5000)

thread = threading.Thread(target=run_app)
thread.daemon = True
thread.start()

# Keep the cell running
import time
while True:
    time.sleep(60)
    print("Server running... (Check the public URL above)")
```
